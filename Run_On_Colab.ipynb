# 🚀 Run Customer Churn Predictor — No Local Install Needed!

This notebook:
1. Clones your project from GitHub
2. Installs all dependencies
3. Starts the Flask backend
4. Exposes it to the internet via **ngrok** — you get a public URL you can open in any browser!

> **Prerequisite:** Get a free ngrok token at https://dashboard.ngrok.com/get-started/your-authtoken

```python
# ── STEP 1: Clone the project from GitHub ──────────────────────────────────
import os

REPO_URL = "https://github.com/dipsur-8536/CUSTOMER_CHURN_PREDICT.git"
REPO_DIR = "CUSTOMER_CHURN_PREDICT"

if os.path.exists(REPO_DIR):
    print("Repo already cloned. Pulling latest changes...")
    os.system(f"git -C {REPO_DIR} pull")
else:
    os.system(f"git clone {REPO_URL}")

os.chdir(REPO_DIR)
print(f"\n✅ Working directory: {os.getcwd()}")
print("Files:", os.listdir('.'))
```

```python
# ── STEP 3: Set your ngrok Auth Token ──────────────────────────────────────
# Get yours FREE at: https://dashboard.ngrok.com/get-started/your-authtoken

NGROK_AUTH_TOKEN = "YOUR_AUTHTOKEN_HERE"   # <-- ✏️ PASTE YOUR TOKEN HERE

from pyngrok import ngrok
ngrok.set_auth_token(NGROK_AUTH_TOKEN)
print("✅ ngrok token set!")
```

```python
# ── STEP 4: Start Flask + open public tunnel ────────────────────────────────
import threading
from app import app

PORT = 5000

# Create a public tunnel to the Flask port
public_url = ngrok.connect(PORT).public_url

print("\n" + "="*65)
print("🎯  YOUR WEB APP IS LIVE! Open this link in your browser:")
print(f"\n       👉  {public_url}  👈")
print("\n" + "="*65)
print("(Keep this cell running to stay online. Re-run to refresh the URL.)")

# Run Flask in a background thread so this cell doesn't block
flask_thread = threading.Thread(
    target=app.run,
    kwargs={"use_reloader": False, "port": PORT, "host": "0.0.0.0"}
)
flask_thread.daemon = True
flask_thread.start()
```
